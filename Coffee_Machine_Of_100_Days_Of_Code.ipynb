Python Coffee Machine Project Requirements
Project Overview
Digital replication of a virtual, coin-operated coffee machine built from scratch in Python (main.py in PyCharm).

1.  Ingredients & Flavors

  - Espresso, Latte, Cappuccino
  - Each drink uses custom quantities of:
      - Water
      - Milk
      - Coffee
  - Each item has an associated recipe price.

2.  Starting Resources

  - Water: 300 ml
  - Milk: 200 ml
  - Coffee: 100 g

3.  Coin Specifications (US Coins)

  - Penny: $0.01
  - Nickel: $0.05
  - Dime: $0.10
  - Quarter: $0.25

Core Requirements

Print Report

  - Triggered by typing report.
  - Displays current remaining quantities of water, milk, coffee, and collected money.

Resource Sufficiency Check

  - When a user orders, verify if ingredients are sufficient.
  - If insufficient, output feedback (e.g., "Sorry, there's not enough water.") and halt the transaction.

Process Coins & Transaction Success

  - Ask user to insert coins and input quantities for each type.
  - Calculate total monetary value.
  - If money is insufficient, refund the cash and cancel the drink.
  - If money is sufficient, calculate change, add profit to machine coffers, and serve the beverage.

Deduct Resources

  - Automatically subtract recipe ingredients from total resources upon successful coffee generation.

```python
MENU = {
    "espresso": {
        "ingredients": {
            "water": 50,
            "coffee": 18,
        },
        "cost": 1.5,
    },
    "latte": {
        "ingredients": {
            "water": 200,
            "milk": 150,
            "coffee": 24,
        },
        "cost": 2.5,
    },
    "cappuccino": {
        "ingredients": {
            "water": 250,
            "milk": 100,
            "coffee": 24,
        },
        "cost": 3.0,
    }
}

profit = 0
resources = {
    "water": 300,
    "milk": 200,
    "coffee": 100,
}
```

```python
turn_off = False

def resources_sufficient(order_ingridents):
  for item in order_ingridents:
    if order_ingridents[item] >= resources[item]:
      print(f"Sorry there is not enough {item}.")
      return False
  return True

def process_coins():
  print("Please insert coins.")
  total = int(input("How many quarters?: ")) * 0.25
  total += int(input("How many dimes?: ")) * 0.1
  total += int(input("How many nickles?: ")) * 0.05
  total += int(input("How many pennies?: ")) * 0.01
  return total

def transaction_successful(money_received, drink_cost):
  if money_received >= drink_cost:
    change = round(money_received - drink_cost, 2)
    print(f"Here is ${change} in change.")
    global profit
    profit += drink_cost
    return True
  else:
    print("Sorry that's not enough money. Money refunded.")
    return False

def make_coffee(drink_name, drink_ingredients):
  for item in drink_ingredients:
    resources[item] -= drink_ingredients[item]
  print(f"Here is your {drink_name} ☕️. Enjoy!")

while True:
  choice = input("What would you like? (espresso/latte/cappuccino): ")
  if choice == "off":
    turn_off = True
    break
  elif choice == "report":
    print(f"Water: {resources['water']}ml")
    print(f"Milk: {resources['milk']}ml")
    print(f"Coffee: {resources['coffee']}g")
    print(f"Money: ${profit}")
  else:
    drink = MENU[choice]
    if resources_sufficient(drink["ingredients"]):
      payment = process_coins()
      if  transaction_successful(payment, drink["cost"]):
        make_coffee(choice, drink["ingredients"])
```
