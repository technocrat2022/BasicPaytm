```python
import datetime
class Transaction:
  def __init__(self, amount, transaction_type, timestamp):
    self.amount = amount
    self.transaction_type = transaction_type
    self.timestamp = timestamp

  def __str__(self):
    return f"[{self.timestamp.strftime('%Y-%m-%d %H:%M:%S')}] {self.transaction_type}: ${self.amount:.2f}"

  def __eq__(self, other):
    if isinstance(other, Transaction):
      return self.amount == other.amount and self.transaction_type == other.transaction_type and self.timestamp == other.timestamp
    return False

class Account:
  def __init__(self, owner_name, balance=0.0, ledger=None):
    self.owner_name = owner_name
    self._balance = 0.0
    self.balance = balance
    if  ledger == None:
      self.ledger = []
    else:
      self.ledger = ledger

  @property
  def  balance(self):
    print("Fetching balance..........")
    return self._balance

  @balance.setter
  def balance(self, new_balance):
    if new_balance < 0.0:
      raise ValueError("Balance cannot be negative.")
    self._balance = new_balance

  def deposit(self, amount):
    self._balance += amount

    self.trans = Transaction(amount, "deposit", datetime.datetime.now())
    self.ledger.append(self.trans)
    print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")

  def withdraw(self, amount):
    if self._balance < amount:
      print("Insufficient funds.")
      return False

    self._balance -= amount
    trans = Transaction(amount, "withdrawal", datetime.datetime.now())
    self.ledger.append(trans)
    print(f"Withdrew ${amount: .2f}. New balance: ${self.balance: .2f}")
    return True

  def flashback_balance(self, steps):
    if steps > len(self.ledger):
      raise ValueError("Invalid number of steps.")

    historical_balance = self.balance
    for trans in self.ledger[-steps:]:
      if trans.transaction_type == "deposite":
        historical_balance -= trans.amount
      elif trans.transaction_type == "withdrawal":
        historical_balance += trans.amount

    print(f"-> Flashback {steps} steps: Historical Balance was ${historical_balance:.2f}")
    return historical_balance

class SavingsAccount(Account):
  def __init__(self, owner_name, interest_rate, balance=0.0, ledger=None):
    super().__init__(owner_name, balance, ledger)
    self.interest_rate = interest_rate

  def apply_interest(self):
    amount = self.balance * self.interest_rate
    self.deposite(amount)
    return self.balance


# --- Test Execution ---
user_1 = Account('Alice')
user_1.deposit(1000)
user_1.withdraw(200)
user_1.deposit(500)
user_1.withdraw(5000)      # Fails safely
user_1.flashback_balance(2) # Temporarily rolls back the +500 and -200
print(f"Live Balance Check: ${user_1.balance:.2f}") # Proves live balance is intact
```
